```python
# Celda 0.4 · Conectarse a la API pública de XM y explorar su catálogo de métricas
# XM es el operador del sistema eléctrico colombiano. Su librería oficial (pydataxm) consulta la API por nosotros.
from pydataxm.pydataxm import ReadDB

try:
    api = ReadDB()
    catalogo = api.get_collections()
    print("Conexión con XM lista ✔  ·  métricas en el catálogo:", len(catalogo))
except Exception as e:
    print("❗ No fue posible conectarse con XM:", e)
    print("   Espere un par de minutos y vuelva a ejecutar esta celda (el servidor a veces está ocupado).")

def buscar(palabra):
    """Devuelve las filas del catálogo de XM que contienen la palabra (sin importar mayúsculas)."""
    coincide = catalogo.astype(str).apply(lambda col: col.str.contains(palabra, case=False, na=False))
    columnas = [c for c in ["MetricId", "MetricName", "Entity", "MetricUnits", "MaxDays"] if c in catalogo.columns]
    return catalogo.loc[coincide.any(axis=1), columnas]

buscar("Precio")
```

```python
# Celda 0.5 · Una función para descargar cualquier métrica de XM como serie DIARIA
def solicitar_datos_xm(metrica, entidad, inicio, fin):
    """Consulta XM con requests síncrono para evitar incompatibilidad asyncio en Python 3.14."""
    tipos = api.inventario_metricas.query("MetricId == @metrica and Entity == @entidad")
    if tipos.empty:
        raise ValueError(f"XM no reconoce la métrica {metrica} para la entidad {entidad}")

    tipo = tipos["Type"].iloc[0]
    if tipo == "ListsEntities":
        respuesta = requests.post(
            "https://servapibi.xm.com.co/lists",
            json={"MetricId": metrica, "Entity": entidad}, timeout=90
        )
        respuesta.raise_for_status()
        datos = respuesta.json()
        return pd.json_normalize(datos["Items"], "ListEntities", "Date", sep="_")

    endpoints = {
        "HourlyEntities": ("hourly", "HourlyEntities"),
        "DailyEntities": ("daily", "DailyEntities"),
        "MonthlyEntities": ("monthly", "MonthlyEntities"),
        "AnnualEntities": ("annual", "AnnualEntities"),
    }
    if tipo not in endpoints:
        raise ValueError(f"Tipo de métrica XM no compatible: {tipo}")
    periodo, endpoint = endpoints[tipo]
    url = f"https://servapibi.xm.com.co/{periodo}"
    fecha = pd.Timestamp(inicio)
    fin = pd.Timestamp(fin)
    tablas = []
    while fecha <= fin:
        fin_mes = min(fecha + pd.offsets.MonthEnd(0), fin)
        cuerpo = {
            "MetricId": metrica,
            "StartDate": fecha.strftime("%Y-%m-%d"),
            "EndDate": fin_mes.strftime("%Y-%m-%d"),
            "Entity": entidad,
            "Filter": [],
        }
        respuesta = requests.post(url, json=cuerpo, timeout=90)
        respuesta.raise_for_status()
        datos = respuesta.json()
        tablas.append(pd.json_normalize(datos["Items"], endpoint, "Date", sep="_"))
        fecha = fin_mes + pd.Timedelta(days=1)
    return pd.concat(tablas, ignore_index=True) if tablas else pd.DataFrame()


def descargar_diario(metrica, entidad, inicio, fin, como="sum"):
    """Descarga una métrica de XM y la devuelve como una serie diaria de pandas.

    metrica : identificador de XM (por ejemplo "DemaCome")
    entidad : "Sistema" (total nacional), "Embalse", "Rio", ...
    como    : "sum"  -> para ENERGÍA (kWh): se suman las 24 horas del día
              "mean" -> para PRECIOS (COP/kWh): se promedian las 24 horas del día
    """
    inicio = pd.to_datetime(inicio).date()
    fin = pd.to_datetime(fin).date()
    tabla = solicitar_datos_xm(metrica, entidad, inicio, fin)        # 1) pedir los datos a la API
    if tabla is None or len(tabla) == 0:                             # 2) ¿llegó algo?
        raise ValueError(f"XM no devolvió datos para {metrica} / {entidad}")
    col_fecha = [c for c in tabla.columns if c.lower() == "date"][0]
    col_horas = [c for c in tabla.columns if "Hour" in c]            # Values_Hour01 ... Values_Hour24
    if col_horas:                                                    # 3a) métrica HORARIA
        horas = tabla[col_horas].apply(pd.to_numeric, errors="coerce")
        tabla["valor"] = horas.sum(axis=1, min_count=1) if como == "sum" else horas.mean(axis=1)
    else:                                                            # 3b) métrica DIARIA ("Value" o "Values_Value")
        col_valor = [c for c in tabla.columns if c.endswith("Value")][0]
        tabla["valor"] = pd.to_numeric(tabla[col_valor], errors="coerce")
    tabla["fecha"] = pd.to_datetime(tabla[col_fecha])
    grupos = tabla.groupby("fecha")["valor"]                         # 4) una fila por día
    serie = grupos.sum(min_count=1) if como == "sum" else grupos.mean()
    return serie.sort_index()

def descargar_con_respaldo(metrica, inicio, fin, como="sum"):
    """Intenta con la entidad 'Sistema' (total nacional); si no existe, suma todos los embalses."""
    try:
        return descargar_diario(metrica, "Sistema", inicio, fin, como)
    except Exception:
        print(f"   '{metrica}' no está disponible como total 'Sistema': se suman los embalses uno por uno.")
        return descargar_diario(metrica, "Embalse", inicio, fin, "sum")

print("Funciones de descarga listas ✔")
```

```python
# Celda 0.6 · Descargar las cuatro series desde 2022 (o leerlas del archivo si ya se descargaron)
INICIO = "2022-01-01"
FIN = (pd.Timestamp.today() - pd.Timedelta(days=1)).strftime("%Y-%m-%d")   # hasta ayer
ARCHIVO = CARPETA_DATOS / "xm_diario.csv"
ACTUALIZAR = False       # cámbielo a True si quiere volver a descargar todo de la API

if ARCHIVO.exists() and not ACTUALIZAR:
    df = pd.read_csv(ARCHIVO, parse_dates=["fecha"], index_col="fecha")
    print("Datos leídos del archivo guardado:", ARCHIVO)
else:
    print("Descargando de la API de XM entre", INICIO, "y", FIN, "(2 a 5 minutos)...")
    demanda = descargar_diario("DemaCome", "Sistema", INICIO, FIN, como="sum")  / 1e6    # kWh → GWh
    print("   ✔ demanda")
    precio = descargar_diario("PrecBolsNaci", "Sistema", INICIO, FIN, como="mean")        # COP/kWh
    print("   ✔ precio de bolsa")
    volumen = descargar_con_respaldo("VoluUtilDiarEner", INICIO, FIN) / 1e6               # kWh → GWh
    print("   ✔ volumen útil")
    capacidad = descargar_con_respaldo("CapaUtilDiarEner", INICIO, FIN) / 1e6             # kWh → GWh
    print("   ✔ capacidad útil")
    df = pd.concat({"demanda_gwh": demanda, "precio_bolsa": precio,
                    "volumen_gwh": volumen, "capacidad_gwh": capacidad}, axis=1, sort=True)
    df.index.name = "fecha"
    df["pct_embalse"] = 100 * df["volumen_gwh"] / df["capacidad_gwh"]   # el porcentaje se RECALCULA
    df.to_csv(ARCHIVO)
    print("Guardado en", ARCHIVO)

print("Filas:", len(df), "· desde", df.index.min().date(), "hasta", df.index.max().date())
df.tail()
```

📖 **Glosario de la misión:** `shape` = (filas, columnas) · `dtypes` = tipo de cada columna · `isna()` = ¿falta el dato? · `date_range(..., freq="D")` = lista de fechas día a día (`"W"` sería semana a semana) · `dropna()` = quita filas con datos faltantes · `.min()` / `.max()` = el menor / el mayor · **rezago de publicación** = días de retraso con que XM publica una serie.

```python
# Celda 1.1 · Perfil de calidad de una tabla con índice de fechas diarias
def perfil_calidad(tabla):
    """Imprime un diagnóstico de calidad y devuelve sus resultados en un diccionario."""
    print("Filas y columnas:", tabla.shape)
    print("Primera fecha:", tabla.index.min().date(), "| Última fecha:", tabla.index.max().date())
    print("\nTipo de dato de cada columna:")
    print(tabla.dtypes.to_string())

    nulos = tabla.isna().sum
    print("\nDatos faltantes (NaN) por columna:")
    print(nulos)

    duplicadas = int(tabla.index.duplicated().sum())
    print("\nFechas duplicadas:", duplicadas)

    calendario = pd.date_range(tabla.index.min(), tabla.index.max(), freq="W")   # todos los días esperados
    faltantes = calendario.difference(tabla.index)
    print("Días esperados:", len(calendario), "| Días que faltan en el calendario:", len(faltantes))

    print("\nResumen numérico:")
    print(tabla.describe().T[["min", "max", "mean"]])
    return {"nulos": nulos, "duplicadas": duplicadas, "esperados": len(calendario), "dias_faltantes": len(faltantes)}

diagnostico = perfil_calidad(df)
```

```python
# Celda 7.1 · Dos funciones propias
def resumen_serie(serie):
    """Devuelve un diccionario con mínimo, máximo, promedio y las fechas del mínimo y del máximo."""
    return {
        "minimo": round(serie.min(), 2),
        "maximo": round(serie.max(), 2),
        "promedio": round(serie.mean(), 2),
        "fecha_minimo": serie.idxmax().date(),
        "fecha_maximo": serie.idxmax().date(),
    }

def semaforo_embalse(pct):
    """Clasifica el nivel de los embalses.
    Devuelve 'sin dato' si pct falta (NaN), 'crítico' si pct < 40,
    'vigilancia' si 40 <= pct < 80 y 'seguro' si pct >= 80."""
    if pct < 80:
        return "vigilancia"
    elif pct < 40:
        return "crítico"
    else:
        return "seguro"

print("Funciones definidas ✔")
```
